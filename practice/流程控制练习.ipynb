# A 档 · if / elif / else 分支

判断类题目，核心是**从上往下依次判断**，命中一个后面就不看了。

## 题1 [A] 奇偶判断

**要求**：让用户输入一个整数，如果是偶数输出“偶数”，否则输出“奇数”。

**提示**：用取余 `%`：`n % 2 == 0` 就表示能被 2 整除。

````python
# 题1 你的代码写在这里

num=int(input("请输入一个整数: "))
if num%2==0:
    print(f"{num}是偶数")
else:
    print(f"{num}是奇数")
````

## 题2 [A] 成绩等级

**要求**：输入一个 0~100 的分数，输出等级：90 及以上 → A，80~89 → B，70~79 → C，60~69 → D，60 以下 → E。

另外：分数小于 0 或大于 100 时，输出“分数不合法”。

**提示**：从上往下写：先 `if score >= 90`，再 `elif score >= 80`…… 想一想为什么不用写 `80 <= score < 90` 也能对。

````python
# 题2 你的代码写在这里

score=int(input("请输入你的成绩: "))
if score>=90:
    print("A")
elif score>=80:
    print("B")
elif score>=70:
    print("C")
elif score>=60:
    print("D")
else:
    print("E")
````

## 题3 [A] 闰年判断

**要求**：输入一个年份，是闰年输出“闰年”，否则输出“平年”。

规则：能被 4 整除但不能被 100 整除，**或者**能被 400 整除。

**提示**：`(year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)`。拿 2000（闰）、1900（平）、2024（闰）、2023（平）各测一遍。

````python
# 题3 你的代码写在这里

year=int(input("请输入年份: "))
if (year%4==0 and year%100!=0) or (year%400==0):
    print(f"{year}是闰年")
else:
    print(f"{year}是平年")
````

## 题4 [A] 三个数里找最大

**要求**：输入三个整数 a、b、c，输出最大的那个。**不许用 `max()`**。

**提示**：先假设 a 最大，再拿 b 和 c 逐个来“挑战”它：`if b > big: big = b`。

````python
# 题4 你的代码写在这里
num_1=int(input("请输入第一个整数: "))
num_2=int(input("请输入第二个整数: "))
num_3=int(input("请输入第三个整数: "))
max=num_1 if num_1>num_2 else num_2
max=num_3 if max<num_3 else max
print(f"{num_1}\t{num_2}\t{num_3}\n最大数为{max}")
````

循环类题目，核心是**累加变量初始 0、累乘变量初始 1**，`while` 里别忘了让条件趋于结束。

## 题5 [B] 1~100 求和

## 题7 [B] 求阶乘

**要求**：输入一个非负整数 n，求 n! = 1×2×3×…×n 并输出（规定 0! = 1，1! = 1）。

**提示**：累乘变量初始值必须是 **1**，写成 0 的话全部乘成 0。

````python
# 题7 你的代码写在这里
n=int(input("请输入一个整数:"))
N=1
if n==0 or n==1:
    print(f'{n}!=1')
else:
    for i in range(1,n+1):
        N*=i
    print(f'{n}!={N}')
````

## 题8 [B] 打印九九乘法表

**要求**：用**嵌套 for 循环**打印九九乘法表（下三角形状）：

```
1×1=1
1×2=2   2×2=4
1×3=3   2×3=6   3×3=9
…共 9 行
```

**提示**：外层 i 从 1 到 9，内层 j 从 1 到 i；`print(f"{j}×{i}={i*j}", end="\t")` 让内容不换行；一行打完后单独 `print()` 换行。

````python
# 题8 你的代码写在这里
for i in range(1,10):
    for j in range(1,i+1):
        print(f'{j}x{i}={i*j}',end="\t")
    print(end='\n')
````

# C 档 · 循环 + 条件，break / continue

`break` 是「整个循环不要了」，`continue` 是「这一轮不要了，继续下一轮」。

## 题10 [C] 判断素数

**要求**：输入一个大于 1 的整数 n，判断它是不是素数（质数：只能被 1 和它自己整除），是就输出“是素数”，否则输出“不是素数”。

**提示**：循环 i 从 2 开始试，一旦 `n % i == 0` 就说明不是素数，用 `break` 提前结束。

小技巧：其实只要试到 `int(n ** 0.5)` 就够了，想想为什么。

````python
# 题10 你的代码写在这里
n=int(input('请输入一个整数:'))
a=True
for i in range(2,n):
    if n%i==0:
        a=False
        break   
b="素数"if a else "非素数"
print(f'{n}是{b}')
````

## 题11 [C] 跳过数字（continue）

**要求**：用 `continue` 打印 1~100 中所有**既不是 7 的倍数、也不含数字 7** 的数。

**提示**：判断含不含 7 可以把数字转成字符串：`'7' in str(i)`。

````python
# 题11 你的代码写在这里
for i in range(1,101):
    if i%7==0 or '7'in str(i):
        continue
    print(i,end='\t')
````

## 题14 [C] 统计字符类型

**要求**：给一个字符串，统计其中英文字母、数字、空格、其他字符各有多少个。

自己造一个字符串来测，例如 `"Hello 2026, 你好 Python!"`。

**提示**：遍历 `for ch in text`；`'a' <= ch <= 'z' or 'A' <= ch <= 'Z'` 判断英文字母；`ch.isdigit()` 判断数字。

注意：`ch.isalpha()` 会把中文也算成字母，别用错了。

````python
# 题14 你的代码写在这里
s="Hello 2026, 你好 Python!"
english=number=space=other=0
for ch in s:
    if ('a'<=ch<='z') or ('A'<=ch<='Z'):
        english += 1
    elif '0'<=ch<='9':
        number += 1
    elif ch==' ':
        space += 1
    else:
        other += 1
print('字母数量:',english)
print('数字数量:',number)
print('空格数量:',space)
print('其他字符数量:',other)
````

# D 档 · 综合挑战

想不出公式时，**穷举**是最稳的解法。慢慢来。

## 题15 [D] 猜数字游戏

**要求**：程序随机生成一个 1~100 的整数（`import random`，`random.randint(1, 100)`），让用户反复猜：

- 猜大了 → 提示“太大了”
- 猜小了 → 提示“太小了”
- 猜对了 → 输出“猜对了！你一共猜了 N 次”并结束

再加一条规则：**最多猜 7 次**，超过 7 次还没中，就输出答案并结束。

**提示**：`while True:` 配合 `break`；用一个 `count` 变量记录次数。

````python
# 题15 你的代码写在这里
#随机一个数
import random
randnum=random.randint(1,100)
#定义count变量
count=0
#循环
while True:
#判断次数    
    if count!=7:
        num=int(input('请输入您猜的数字:'))
    else:
        print('已使用七次机会,游戏结束')
        break
#判断数的大小
    if num>randnum:
        print('太大了\n请继续输入')
        count+=1
        continue
    elif num<randnum:
        print('太小了\n请继续输入')
        count+=1
        continue
    else:
        print('猜对了')
        count+=1
        break
#输出随机数结果
print(f'结果是{randnum}')
````

## 题16 [D] 求平均分（个数不定）

**要求**：反复输入成绩，当输入 0 时结束（0 不算成绩）；结束后输出：一共输入了几个成绩、总分是多少、平均分是多少。

注意：如果一个成绩都没输入，要提示“没有输入成绩”，不能直接算平均分。

**提示**：`while True` + `input` + `if score == 0: break`。想一想为什么没有任何成绩时算平均分会报错。

````python
# 题16 你的代码写在这里
s=[]
while True:
    score=float(input('请输入成绩:'))
    if score==0:
        break
    s.append(score)
if len(s)==0:
    print('没有输入成绩')
else:
    print(s)
    print('成绩个数:',len(s))
    print('总分:',sum(s))
    print('平均分:',sum(s)/len(s))
````

## 题17 [D] 鸡兔同笼

**要求**：输入头的数量 heads 和脚的数量 feet，用**循环穷举鸡的只数**，求出鸡和兔各多少只。如果怎么试都对不上，输出“无解”。

例：heads=35，feet=94 → 鸡 23 只，兔 12 只。

**提示**：`for chicken in range(heads + 1):` 里面 `rabbit = heads - chicken`，判断 `2 * chicken + 4 * rabbit == feet`。

````python
# 题17 你的代码写在这里
heads=int(input('输入头的数量 heads:'))
feet=int(input('输入脚的数量 feet:'))
found=True
for i in range(heads+1):
    j=heads-i
    if 2*i+4*j==feet:
        print(f'鸡的数量是:{i},兔的数量是:{j}')
        found=False
if found:
    print('无解')
````
