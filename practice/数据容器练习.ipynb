# 五种数据容器 · 特点总表

| 容器 | 语法 | 有序？ | 可变？ | 元素唯一？ | 取值方式 | 一句话 |
|---|---|---|---|---|---|---|
| **list 列表** | `[1, 2, 3]` | ✅ 有序 | ✅ 可变 | ❌ 可重复 | 下标 `a[0]` | 能改的序列 |
| **tuple 元组** | `(1, 2, 3)` | ✅ 有序 | ❌ 不可变 | ❌ 可重复 | 下标 `a[0]` | 不能改的序列 |
| **str 字符串** | `"abc"` | ✅ 有序 | ❌ 不可变 | ❌ 可重复 | 下标 `a[0]` | 一堆字符的序列 |
| **set 集合** | `{1, 2, 3}` | ❌ 无序 | ✅ 可变 | ✅ **自动去重** | 不能下标，只能 `in` | 去重 + 集合运算 |
| **dict 字典** | `{"a": 1}` | ✅ 插入有序 | ✅ 可变 | 键唯一 | 键 `a["k"]` | 键值对映射 |

## 最省脑子的记法：它们分两派

**序列派：list / tuple / str**
- 有**顺序**，能 `[下标]` 取值、能**切片** `a[1:3]`
- 支持 `+` 拼接、`*` 重复、`in`、`len()`、`min/max`、`sorted()`、`count()`、`index()`
- 查一个元素要**从头遍历** → `in` 是 O(n)，数据多了就慢
- 三者的区别只在**能不能改**：list 能改；tuple 和 str 不能改

**哈希派：set / dict**
- **没有下标、不能切片**，写 `a[0]` 直接报错
- 靠**哈希表**存储 → 查/增/删平均 **O(1)**，`in` 判断极快，这是它们存在的最大理由
- set 存「单个元素」，dict 存「键 → 值」

> 记牢一句话：**要顺序用序列，要快和去重用哈希。**

## 题3 [A] 遍历求总分 / 最高分 / 平均分

**要求**：给成绩列表 `[88, 92, 79, 95, 61]`：

1. 用**循环**求总分、最高分、平均分
2. 再用内置函数 `sum()` / `max()` 写一遍，对比结果是否一致

**提示**：总分：累加变量初始 0；最高分：**打擂台**（先假设第一个最大，再逐个挑战 `if x > max_score: max_score = x`）；平均分 = 总分 / 个数。

````python
# 题3 你的代码写在这里
list=[88, 92, 79, 95, 61]
sum_num=0
max_num=0
count=0
for i in list:
    count+=1
    sum_num+=i
    if i>=max_num:
        max_num=i
print(sum_num)
print(max_num)
print(sum_num/count)

print(sum(list))
print(max(list))
print(sum(list)/len(list))
````
