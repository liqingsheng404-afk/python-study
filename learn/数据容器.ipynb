```python
num1 = float(input("请输入第一个数字："))
operator = input("请输入运算符（+、-、*、/）：")
num2 = float(input("请输入第二个数字："))

if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    result = "除数不能为零" if num2 == 0 else num1 / num2
else:
    result = "无效的运算符"

print("结果：", result)
```

```python
s=[1, 2, 3, 'hello', 4, 5,58,True,False]
#末尾追加元素   append()
s.append('world')
print(s)
#在指定位置之前插入元素   insert()
s.insert(6, 'python')
print(s)
#删除第一个匹配到的元素   remove()
s.remove(False)
print(s)
#删除指定索引位置的元素   pop()
s.pop(4)
print(s)
#反转列表   reverse()
s.reverse()
print(s)

s.remove('hello')
s.remove('world')
s.remove('python')
s.pop(0)
print(s)

#排序   sort()
s.sort()
print(s)
```

列表推导式

```python
#生成1-20的平方列表
# a=[]
# for i in range(1,21):
#     j=i**2
#     a.append(j)
# print(a)

#列表推导式-->s=[要插入的值 / for i in 列表]
a=[i**2 for i in range(1,21)]
print(a)

#提取列表中的偶数,并计算其平方
# b=[]
# for i in a:
#     if i%2==0:
#         j=i**2
#         b.append(j)
# print(b)
#列表推导式-->s=[要插入的值 / for i in 列表 / if 条件]
b=[i**2 for i in a if i%2==0]
print(b)
```

```python
s='  hello world , hello python  '
print(s.find('world'))#查找字符串中第一次出现的位置
print(s.count('o'))#统计字符串中字符出现的次数
print(s.upper())#将字符串中的字母全部转换为大写
print(s.lower())#将字符串中的字母全部转换为小写
print(s.split(' '))#将字符串按照指定的分隔符进行分割
print(s.strip())#去除字符串两端的空格
print(s.replace('hello','hi'))#将字符串中的字符进行替换
print(s.startswith(' '))#判断字符串是否以指定的字符开头
print(s.endswith('*'))#判断字符串是否以指定的字符结尾
print(s)
```
